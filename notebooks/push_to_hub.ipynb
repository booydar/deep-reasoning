```python
import torch
from modeling_amt.language_modeling import AssociativeRecurrentWrapper, AssociativeMemoryCell
from transformers import AutoModelForCausalLM

base_model = AutoModelForCausalLM.from_pretrained('gpt2')
memory_cell = AssociativeMemoryCell(base_model, num_mem_tokens=16, d_mem=32, correction=False, wrap_pos=False)
rmt = AssociativeRecurrentWrapper(memory_cell, segment_size=1024)

with open(cpt_path, "rb") as f:
    cpt = torch.load(f)

rmt.load_state_dict(cpt)
```

```python
from huggingface_hub import HfApi, create_repo, upload_file

model_name = "booydar/ARMT-gpt2-mem16d32-multiplication_4x4"
cpt_path = "/workspace-SR006.nfs2/bulatov/rmt/runs/multiplication_4x4/gpt2-armt/1x1024_mem16_1024_LR1e-03-cot/checkpoint-21000/pytorch_model.bin"


# Upload your bin file
upload_file(
    path_or_fileobj=cpt_path,
    path_in_repo="pytorch_model.bin",
    repo_id=model_name,
    commit_message="Add model weights",
    token=os.environ['hf_token']
)
```

```python
model_name = "booydar/ARMT-gpt2-mem16d32-multiplication_4x4"
rmt.push_to_hub(model_name, token=os.environ['hf_token'], safe_serialization=False)
```
